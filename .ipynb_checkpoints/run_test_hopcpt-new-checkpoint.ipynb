```python


def compute_hopfield_net_loss_(hopfield_net, memory_feature, query_feature, memory_value, target_value):
    """
    :param context_x: (batch_size, memory_length, dim_feature)
    :param query_x: (batch_size, 1, dim_feature)
    :param target: (batch_size, 1, 1)
    """
    if target_value.dim() == 2:
        target_value.unsqueeze_(-1) 

    preds = hopfield_net(memory_feature, query_feature, memory_value) # (batch_size, 1, 1)
    e_abs = target_value.abs()
    loss = ((e_abs - preds) ** 2).mean(dim=1)   # average over time (T^{-1} sum)
    loss = loss.mean()                 # average over batch

    return loss
```
