```python
def generate_context_hopcpt(strided_x, strided_y, target_x, target_predictions, features_used):
    """
    :param strided_x: strided x for memory context, (batch_size, memory_len, dim)
    :param strided_y: strided y for memory context, (batch_size, memory_len, 1)
    :param target_x: target x for query context, (batch_size, 1, dim)
    :param target_predictions: target \hat{y} for query context, (batch_size, 1, 1)
    """

    device = strided_x.device

    if strided_y.ndim == 2:
        strided_y.unsqueeze_(-1)

    if target_x.ndim == 2:
        target_x.unsqueeze_(-1)

    if target_predictions.ndim == 2:
        target_predictions.unsqueeze_(-1)

    if features_used == "xy":
        strided_feature = torch.cat([strided_x, strided_y], dim=-1).to(device)
        query_feature = torch.cat([target_x, target_predictions], dim=-1).to(device)

    elif features_used == "x":
        strided_feature = strided_x
        query_feature = target_x

    return strided_feature, query_feature
```
